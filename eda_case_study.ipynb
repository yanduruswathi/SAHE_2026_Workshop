```python
#datavisualization -->matplotlib,seaborn,plotly
import matplotlib.pyplot as plt #!pip install matplotlib
```

```python
#distribution of age
counts,edges,bars=plt.hist(data['Age'],bins=15,color='pink')
print(counts)
print(bars)
plt.xlabel("Age")
plt.ylabel("Count")
plt.bar_label(bars)
plt.title("Age distribution")
plt.show()
```

```python
data['Survived'].value_counts()
```

```python
index=data['Survived'].value_counts().keys().tolist()
print(index)
labels=data['Survived'].value_counts().values.tolist()
print(labels)
```

```python
#pie chart
plt.pie(x=labels,
        labels=['Dead','Alive'],
        colors=['red','yellow'],
        autopct="%.1f%%",
        shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#pie chart
plt.pie(x=[577,314],
        labels=['M','F'],
        colors=['red','pink'],
        autopct="%.1f%%",
        shadow=True)
plt.title("Survived Distribution")
plt.legend()
plt.show()
```

```python
#pair-wise relationship -->scatterplot()
plt.scatter(x=data['Age'],
            y=data['Fare'])
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare Distribution")
plt.show()
```

```python
import plotly as py
import plotly.graph_objects as go #for interactive graphs
```

```python
#lets visualize histogram
trace=go.Histogram(x=data['Age'],
                   nbinsx=20,
                   marker={'color':'red'})
#layout define for the graph
layout=go.Layout(title="Age Distribution")
#we combinethe trace  with the layout to display the figure
fig=go.Figure(data=trace,layout=layout)
fig
```

```python
#Now we will display for piechart
trace = go.Pie(labels=['Dead','Alive'],
               values=[549,342],
              )
layout = go.Layout(title = 'Survived Distribution')
fig = go.Figure(data=trace, layout=layout)
fig.show()
```

```python
trace=go.Box(x=data['Pclass'],
             y=data['Age'],
             marker={'color':'green'})
layout=go.Layout(title="Age vs pcalss Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```

```python
trace=go.Scatter(x=data['Age'],y=data['Fare'],
                 mode='markers',
                 marker={'color':'purple'})
layout=go.Layout(
    title="Age vs Fare Distribution")
fig=go.Figure(data=trace,layout=layout)
fig.show()
```
