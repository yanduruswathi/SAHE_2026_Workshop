```python
##step1: importing the libaries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder , PolynomialFeatures
from sklearn.metrics import mean_absolute_error , mean_squared_error,r2_score
sns.set_style(style='whitegrid')
```

```python
#step2:loading and cleaning the data 
data=fetch_ucirepo(id = 9)
data
```

```python
x = data.data.features.copy()
x
```

```python
y=data.data.targets['mpg']
y
```

```python
data
```

```python
x.describe()
x.info()
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python
#first we will transform the horsepower and origin columns
x['horsepower'] = pd.to_numeric(x['horsepower'],errors='coerce')
x['horsepower'].info()
x['origin']=x['origin'].astype(str)
```

```python
#x.info()
#all numerical values into a single collection
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features=['origin']
x=x[numerical_features+categorical_features]
#x.info()
#x.isnull().sum()
x.shape
```

```python
#step 3: quick exploratory data anaylsis (seaborn)
df=x.copy()
df['mpg']=y
```

```python
#df
#we are gonna use sumplots() concept to divide the plot into multiple plots
fig,axes=plt.subplots(1,3,figsize=(16,4))
#plot graphs as per axes
sns.histplot(df['mpg'],kde=False,ax=axes[0],color='red')
#in similar way we display other graphs in rmng plots
axes[0].set_title("MPG Distribution")
sns.scatterplot(x=df['weight'],y=df['mpg'],ax=axes[1])
axes[1].set_title("Weight vs MPG Distribution")
sns.scatterplot(x=df['horsepower'],y=df['mpg'],ax=axes[2],color='yellow')
axes[2].set_title("HoresePower vs MPG Distribution")
```

```python
#correlation beteen our variables (positive coreelation/negative correlation)
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),
            annot=True,
            cmap='coolwarm',
            center=0
           )
plt.title("Correlation matrix")
plt.tight_layout()
plt.show()
```

```python
#Step 4: Dividing the data into train test split
#print(x)
#print(y)
#split rule--> 70-30,80-20,75-25
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print('\nTrain test Split')
print("Training Data", len(x_train))
print("Testing Data", len(y_test))
```

```python
#step 5 :creating an evalutations function for baseline model
def evaluate(model_name,actual,predicted):
    """Returns MAE,MSE,RMSE,R2-Socre"""
    mse=mean_squared_error(actual,predicted)
    mae=mean_absolute_error(actual,predicted)
    return {'Model_Name':model_name,
    'MAE':mae,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-Score':r2_score(actual,predicted)}
results=[]
```

```python
#step6: Baseline model --> it predictsthe training set for every values
baseline_predictions = np.full(len(y_test),y_train.mean())
#baseline_predications
#in above case we are keeping a baseline predication
results.append(evaluate("MEan baseline",y_test,baseline_predictions))
results
```

```python
# step7: we apply linear regression
simple_model = LinearRegression()
#we fit the data the model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions = simple_model.predict(x_test[['weight']])
#print (simple_model_predictions
results.append(evaluate("Simple Linear Regression:Weight",
                        y_test,
                        simple_model_predictions))
print(len(results))
```

results

```python
# we will check the intercept value
print(f'Simple Linear Regression Equation')
print(f'Predicted MPG  = { simple_model.intercept_}=({simple_model.coef_[0]}*weight')
```
